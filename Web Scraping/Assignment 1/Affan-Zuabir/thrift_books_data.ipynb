```python
import re 
import requests
from bs4 import BeautifulSoup 
import csv
import json
```

```python
MAX_PAGES = 22
```

```python
with open('thrift_books_data.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['title', 'condition', 'buy_price'])

    for page in range(1, MAX_PAGES + 1):
        print(f"Scraping page {page}...")

        url = f'https://www.thriftbooks.com/browse/?b.search=&b.page={page}&b.f.t[]=13362'
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')

        scripts = soup.find_all('script')

        if len(scripts) < 13 or scripts[12].string is None:
            print(" Stopping at page", page)
            break

        script_str = scripts[12].string

        match = re.search(r'window\.searchStoreV2\s*=\s*(\{.*?\});', script_str, re.DOTALL)
        if not match:
            print("⚠️ JSON match not found. Stopping at page", page)
            break

        try:
            works_json = json.loads(match.group(1))
            works = works_json.get('works', [])
        except Exception as e:
            print(f" failed at page {page}: {e}")
            break


        for i in works:
            title = i.get('title', 'N/A')
            condition = i.get('buyNowCondition', 'N/A')
            price = i.get('buyNowPrice', 'N/A')
            writer.writerow([title, condition, price])

print("Done!")
```
